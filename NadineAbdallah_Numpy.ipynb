## 1. Introduction & Why NumPy Matters

### 🌍 Real-World Use Case
Imagine you work at an e-commerce company and need to apply a **10% discount** to **1,000,000 product prices**.  
With a standard Python `list`, you have to loop through every single price one by one. With **NumPy**, you multiply the entire array by `0.9` in a single operation — and because NumPy runs compiled C code on contiguous blocks of memory, it is often **10x to 100x faster** and uses much less RAM.

### Python `list` vs. NumPy `ndarray`

| Feature | Python `list` | NumPy `ndarray` |
| :--- | :--- | :--- |
| **Data types** | Mixed types allowed (`[1, "hi", 3.5]`) | Single uniform type (e.g., all `float64`) |
| **Memory layout** | Scattered pointers in memory | Contiguous block of memory (compact & fast) |
| **Math operations** | Requires `for` loops | **Vectorized** (operates on whole arrays at once) |
| **Best used for** | General-purpose collections | Numerical computing, EDA, images, ML matrices |

````python
import numpy as np
import time

print("NumPy version:", np.__version__)

# Quick speed comparison: squaring 1,000,000 numbers
size = 1_000_000
py_list = list(range(size))
np_arr = np.arange(size)

# 1) Python list approach
start = time.perf_counter()
py_result = [x ** 2 for x in py_list]
py_time = time.perf_counter() - start

# 2) NumPy vectorized approach
start = time.perf_counter()
np_result = np_arr ** 2
np_time = time.perf_counter() - start

print(f"Python list time: {py_time:.4f} seconds")
print(f"NumPy array time: {np_time:.4f} seconds")
print(f"Speedup: ~{py_time / np_time:.1f}x faster!")
````

## 2. Creating NumPy Arrays

In NumPy, everything revolves around the **`ndarray`** (N-dimensional array). Depending on the data you are analyzing, your array will usually be **1D**, **2D**, or **3D**:

- **1D Array (Vector):** Like a single column in Excel — e.g., daily temperatures for 7 days, or prices of 5 items.
- **2D Array (Matrix):** Like a spreadsheet or table (**rows × columns**) — e.g., 100 students and their 3 exam scores, or a **grayscale image** (grid of pixel brightness values).
- **3D Array (Tensor):** Like a stack of tables — e.g., a **color RGB image** (height × width × 3 color channels: Red, Green, Blue).

### 2.1 Creating Arrays from Python Lists

````python
# 1D array: daily temperatures (°C) for one week
temps = np.array([22, 24, 19, 25, 27, 26, 23])
print("1D array (temps):", temps)

# 2D array: exam scores for 3 students (rows) across 4 subjects (columns)
scores = np.array([
    [85, 90, 78, 92],
    [70, 88, 95, 80],
    [91, 84, 89, 96]
])
print("\n2D array (scores):\n", scores)

# 3D array: a tiny 2x2 color RGB image (2 rows, 2 cols, 3 color channels)
tiny_image = np.array([
    [[255, 0, 0], [0, 255, 0]],     # Red pixel, Green pixel
    [[0, 0, 255], [255, 255, 0]]    # Blue pixel, Yellow pixel
])
print("\n3D array shape (height, width, channels):", tiny_image.shape)
````

### 2.2 Built-in Array Initializers

#### 🌍 Real-World Use Case
- **`np.zeros()`:** Creating a blank canvas (a black image of all `0` pixels) or an empty table to fill in later during a simulation.
- **`np.ones()` / `np.full()`:** Initializing default weights or filling a scoreboard with a starting value (e.g., everyone starts with `100` points).
- **`np.eye()`:** Creating an **Identity Matrix** (1s on the diagonal, 0s elsewhere), widely used in linear algebra and ML regularization.

````python
# 2x3 matrix of zeros (e.g., blank placeholder table)
zeros_arr = np.zeros((2, 3))
print("Zeros:\n", zeros_arr)

# 2x3 matrix of ones
ones_arr = np.ones((2, 3))
print("\nOnes:\n", ones_arr)

# 2x4 matrix filled with a constant value (e.g., starting score of 100)
full_arr = np.full((2, 4), 100)
print("\nFull (100):\n", full_arr)

# 3x3 Identity matrix
eye_arr = np.eye(3)
print("\nIdentity matrix:\n", eye_arr)
````

### 2.3 Number Sequences: `np.arange()` vs. `np.linspace()`

#### 🌍 Real-World Use Case
- **`np.arange(start, stop, step)`:** Creating day numbers `[1, 2, 3, ..., 30]` for a monthly sales chart (similar to Python's `range()`).
- **`np.linspace(start, stop, num_points)`:** Generating 100 evenly spaced points between `0` and `10` so **Matplotlib** can draw a smooth curve.

> **⚠️ Watch Out:** `np.arange(0, 10, 2)` **excludes** `10` (stop), while `np.linspace(0, 10, 5)` **includes** `10`!

````python
# Days 1 to 14, stepping by 2
days = np.arange(1, 15, 2)
print("arange(1, 15, 2):", days)

# 5 evenly spaced points from 0 to 1 (inclusive)
grid = np.linspace(0, 1, 5)
print("linspace(0, 1, 5):", grid)
````

### 2.4 Random Arrays with `np.random.default_rng()`

#### 🌍 Real-World Use Case
In data science and ML, randomness is everywhere:
- Generating **synthetic datasets** to test your EDA code.
- Randomly **shuffling** rows before splitting into training and test sets.
- Simulating noisy sensor measurements (e.g., Gaussian / normal noise).

Passing a **seed** like `default_rng(42)` ensures **reproducibility** — every time you or a teammate runs the notebook, you get the exact same "random" numbers!

````python
# Create a reproducible random number generator
rng = np.random.default_rng(seed=42)

# Random integers between 50 and 100 (exclusive) -> e.g., simulated test scores
rand_scores = rng.integers(low=50, high=100, size=(3, 4))
print("Random integer scores (3x4):\n", rand_scores)

# Random floats uniformly distributed in [0.0, 1.0)
rand_uniform = rng.random(size=5)
print("\nUniform [0, 1):", np.round(rand_uniform, 3))

# Random numbers from a Normal (bell-curve) distribution (mean=170, std=10 -> heights in cm)
rand_heights = rng.normal(loc=170, scale=10, size=6)
print("\nSimulated heights (cm):", np.round(rand_heights, 1))
````

## 3. Inspecting Arrays & Data Types (`dtype`)

### 🌍 Real-World Use Case
Before cleaning or plotting a dataset, the very first step of **EDA** is inspecting its **shape and types**:
- How many rows (samples) and columns (features) do I have?
- Are values stored as integers (`int64`), decimals (`float64`), or `True`/`False` flags (`bool`)?
- Sometimes ML libraries (or memory limits) require converting booleans `[True, False]` into `[1, 0]`, or shrinking `float64` to `float32` to save half the RAM.

````python
data = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

print("Array:\n", data)
print("Dimensions (ndim):", data.ndim)      # 2D
print("Shape (rows, cols):", data.shape)  # (2, 3)
print("Total elements (size):", data.size)# 6
print("Data type (dtype):", data.dtype)   # int64 (on 64-bit systems)
print("Memory used (bytes):", data.nbytes)
````

### Casting Data Types with `.astype()`
Notice that NumPy arrays can only hold **one data type**. If you mix an integer and a float in a list, NumPy automatically promotes everything to `float64` so no decimal precision is lost. You can also convert types explicitly with `.astype()`.

````python
# Mixing int and float automatically promotes to float64
mixed = np.array([1, 2, 3.5])
print("Mixed array:", mixed, "| dtype:", mixed.dtype)

# Convert floats to integers (truncates decimal part)
as_int = mixed.astype(int)
print("Cast to int:", as_int, "| dtype:", as_int.dtype)

# Convert boolean flags (e.g., 'Purchased?') to 1s and 0s for ML models
purchased = np.array([True, False, True, True])
purchased_numeric = purchased.astype(int)
print("Booleans as 0/1:", purchased_numeric)
````

## 4. Indexing, Slicing & Filtering Data

### 🌍 Real-World Use Case
Suppose you have a dataset of **house prices** or **student exam scores**:
- **Indexing / Slicing:** Grabbing the first 3 rows to preview your data, selecting just the `"Math"` column, or cropping the center of an image.
- **Boolean Masking (Filtering):** Finding *only* the students who scored above 85, or spotting invalid negative ages (`age < 0`) during data cleaning.

### 4.1 1D & 2D Indexing and Slicing
For 2D arrays, use **`arr[row, col]`** (with a comma!). The slicing syntax is **`start:stop:step`** (`stop` is excluded).

````python
# 4 students (rows) x 3 exams: [Math, Science, English]
grades = np.array([
    [88, 92, 85],  # Student 0
    [72, 68, 74],  # Student 1
    [95, 91, 89],  # Student 2
    [60, 75, 70]   # Student 3
])

print("Student 0's Science grade (row 0, col 1):", grades[0, 1])
print("Last student's last grade:", grades[-1, -1])

# Slicing rows and columns:
print("\nFirst 2 students (all subjects):\n", grades[:2, :])
print("\nMath column (all rows, col 0):", grades[:, 0])
print("\nSub-table (rows 1..2, cols 1..2):\n", grades[1:3, 1:3])
````

### ⚠️ Watch Out: Views vs. Copies!
In Python lists, slicing (`my_list[1:3]`) creates a **brand-new copy**.  
In NumPy, slicing creates a **view** (a window into the *same* memory) to save RAM! That means modifying a slice **also modifies the original array** unless you explicitly call **`.copy()`**.

> **🌍 Real-World Use Case:** Imagine slicing out a sample of customer records to test a data-cleaning formula — if you forget `.copy()`, you accidentally overwrite your raw data!

````python
original = np.array([10, 20, 30, 40])

# Slice is a VIEW -> shares memory with `original`
slice_view = original[1:3]
slice_view[0] = 999
print("Original after modifying view:", original)  # Changed!

# Use .copy() when you want an independent copy
original = np.array([10, 20, 30, 40])
safe_copy = original[1:3].copy()
safe_copy[0] = 999
print("Original after modifying .copy():", original)  # Safe!
````

### 4.2 Boolean Masking (Filtering) & Fancy Indexing
Boolean masking is one of the most useful tools in **EDA**. You write a condition (e.g., `ages >= 18`), NumPy creates a `True`/`False` mask, and passing that mask inside `arr[mask]` keeps only the `True` items.

For multiple conditions:
- Use **`&`** for AND (not `and`)
- Use **`|`** for OR (not `or`)
- Use **`~`** for NOT (not `not`)
- Always wrap each condition in parentheses `(...)`!

````python
ages = np.array([15, 22, 34, -5, 19, 120, 28])

# Clean data: find valid adult ages (between 18 and 100)
valid_mask = (ages >= 18) & (ages <= 100)
print("Boolean mask:", valid_mask)
print("Valid adult ages:", ages[valid_mask])

# Find suspicious/invalid values (negative or > 100)
invalid_mask = (ages < 0) | (ages > 100)
print("Suspicious ages:", ages[invalid_mask])

# Fancy indexing: pick specific rows by index list (e.g., Students 0 and 2)
selected_students = grades[[0, 2]]
print("\nSelected students (0 and 2):\n", selected_students)
````

## 5. Reshaping, Transposing & Combining Arrays

### 🌍 Real-World Use Case
- **`.reshape()`:** Scikit-Learn models expect 2D input tables `(n_samples, n_features)`. If you have a 1D list of 6 house sizes, you reshape it to `(6, 1)` before training. Or in computer vision, you flatten a `28×28` image into a 1D vector of `784` numbers!
- **`.T` (Transpose):** Swapping rows and columns when a sensor exports data with timestamps as columns instead of rows.
- **`vstack` / `hstack`:** Appending yesterday's and today's data rows together (`vstack`), or adding a brand-new feature column (`hstack`).

````python
# 1) Reshaping: turn 1D array of 6 house sizes into a 2D column (6 rows, 1 col)
house_sizes = np.array([1200, 1500, 1800, 2000, 2400, 3000])
X = house_sizes.reshape(-1, 1)  # -1 tells NumPy: "figure out this dimension automatically!"
print("Reshaped to (6, 1):\n", X)

# Flatten a 2D array back to 1D
flat = grades.flatten()
print("\nFlattened grades:", flat)

# 2) Transpose (.T): flip rows <-> columns
matrix = np.array([[1, 2, 3],
                   [4, 5, 6]])
print("\nOriginal shape:", matrix.shape)
print("Transposed (.T) shape:", matrix.T.shape)
print(matrix.T)
````

````python
# 3) Combining arrays: vstack (stack vertically = add rows) & hstack (stack horizontally = add cols)
day1 = np.array([[10, 20],
                 [30, 40]])
day2 = np.array([[50, 60]])

all_days = np.vstack([day1, day2])
print("Stacked rows (vstack):\n", all_days)

# Adding a new feature column (must have same number of rows!)
new_col = np.array([[100], [200], [300]])
with_feature = np.hstack([all_days, new_col])
print("\nAdded column (hstack):\n", with_feature)

# 4) Splitting an array into chunks (e.g., splitting 6 samples into 3 batches)
batches = np.split(house_sizes, 3)
print("\nSplit into 3 batches:", batches)
````

## 6. Vectorized Math, Ufuncs & Broadcasting

### 6.1 Element-wise Math & Universal Functions (Ufuncs)

#### 🌍 Real-World Use Case
- **Unit conversion:** Converting an entire column of temperatures from Celsius to Fahrenheit (`C * 9/5 + 32`) in one line.
- **Data transformation in EDA:** House prices and salaries are often heavily right-skewed; data scientists apply **`np.log()`** to make distributions more normal before modeling.

````python
prices = np.array([100, 250, 400, 1000])

# Element-wise arithmetic
discounted = prices * 0.85   # 15% off
with_tax = discounted + 10   # +$10 flat shipping
print("Discounted + $10 shipping:", with_tax)

# Universal functions (ufuncs)
print("Square root:", np.round(np.sqrt(prices), 2))
print("Natural log:", np.round(np.log(prices), 2))
````

### 6.2 Broadcasting (One of NumPy's Superpowers!)

#### 🌍 Real-World Use Case
Suppose you have a table of **3 students × 3 subjects** (`scores`), and you want to **add bonus points** `[+5, +0, +2]` to Math, Science, and English respectively — or subtract the **mean of each column** from every row (a core ML preprocessing step called *centering*!).

Even though `scores` has shape **`(3, 3)`** and `bonus` has shape **`(3,)`**, NumPy automatically **"stretches" (broadcasts)** the smaller array across the rows without copying data in memory:

```text
scores (3, 3)      bonus (3,) -> treated as (1, 3) -> stretched to (3, 3)
[[80, 70, 90],     [[5, 0, 2],
 [60, 85, 75],  +   [5, 0, 2],
 [90, 95, 88]]      [5, 0, 2]]
```

**Simple Rule of Thumb:** Two dimensions are compatible for broadcasting when reading shapes **from right to left**:
1. They are equal, **OR**
2. One of them is `1` (or missing).

````python
scores = np.array([
    [80, 70, 90],
    [60, 85, 75],
    [90, 95, 88]
])
bonus = np.array([5, 0, 2])  # +5 Math, +0 Science, +2 English

updated_scores = scores + bonus
print("Scores after broadcasting bonus:\n", updated_scores)
````

## 7. Aggregations, Summary Statistics & the `axis` Parameter

### 🌍 Real-World Use Case
Summarizing data is the heart of **EDA**:
- What is the **mean** vs. **median** salary? *(Why both? One billionaire skews the mean, while the median stays robust!)*
- What is the **standard deviation** (`std`)? *(How spread out are the numbers?)*
- What are the **25th and 75th percentiles**? *(Used for boxplots and outlier detection!)*

````python
salaries = np.array([45_000, 48_000, 50_000, 52_000, 55_000, 500_000])  # Notice the 500k outlier!

print("Mean salary:  $", round(np.mean(salaries), 2))
print("Median salary:$", np.median(salaries))
print("Std deviation:$", round(np.std(salaries), 2))
print("Min / Max:    $", np.min(salaries), "/ $", np.max(salaries))
print("25th & 75th percentiles:", np.percentile(salaries, [25, 75]))
````

### 7.1 Understanding `axis=0` vs. `axis=1`

Beginners often mix up `axis=0` and `axis=1`. Here is the easiest way to remember it: **the `axis` you specify is the axis that gets *collapsed*!**

- **`axis=0` (collapse rows ↓):** Computes **one result per column** (e.g., *average score per subject*).
- **`axis=1` (collapse columns →):** Computes **one result per row** (e.g., *average score per student*).

````python
# 3 students (rows) x 4 subjects (columns)
exam_table = np.array([
    [80, 90, 70, 85],  # Student 0
    [60, 75, 80, 70],  # Student 1
    [95, 92, 88, 96]   # Student 2
])

print("Overall mean (all numbers):", np.mean(exam_table))
print("Mean per subject (axis=0, collapses rows):", np.mean(exam_table, axis=0))
print("Mean per student (axis=1, collapses cols):", np.mean(exam_table, axis=1))
````

### 7.2 Sorting, `argmax`/`argmin`, and `unique`

#### 🌍 Real-World Use Case
- **`argmax`:** Finding which model achieved the highest accuracy, or which day had peak website traffic.
- **`np.unique(..., return_counts=True)`:** Checking **class balance** in a classification dataset (e.g., how many `"Spam"` vs. `"Not Spam"` emails do we have?).

````python
accuracies = np.array([0.78, 0.91, 0.85, 0.89])
print("Highest accuracy:", np.max(accuracies), "at model index:", np.argmax(accuracies))
print("Indices sorted ascending (argsort):", np.argsort(accuracies))

# Checking class counts (0 = Not Spam, 1 = Spam)
labels = np.array([0, 0, 1, 0, 1, 0, 0, 1])
values, counts = np.unique(labels, return_counts=True)
print("Unique classes:", values, "| Counts:", counts)
````

### 7.3 Handling Missing Values (`np.nan`)

#### 🌍 Real-World Use Case
Real-world datasets are messy — sensors drop readings and survey respondents skip questions. In NumPy, missing numerical values are represented as **`np.nan`** (*Not a Number*).

> **⚠️ Watch Out:** Standard functions like `np.mean()` return `nan` if even **one** value is missing! Always use **`np.isnan()`** to detect missing values and **`np.nanmean()` / `np.nansum()`** to ignore them.

````python
sensor_readings = np.array([21.5, 22.0, np.nan, 23.1, np.nan, 21.8])

print("Number of missing values:", np.isnan(sensor_readings).sum())
print("Regular np.mean():", np.mean(sensor_readings))       # Returns nan!
print("Safe np.nanmean():", round(np.nanmean(sensor_readings), 2))
````

## 8. Conditional Logic & Matrix Multiplication for ML

### 8.1 `np.where()` and `np.clip()`

#### 🌍 Real-World Use Case
- **`np.where(condition, value_if_true, value_if_false)`:** Quickly creating binary labels (`"Pass"` if score `>= 60` else `"Fail"`), or replacing `NaN` values with the column average!
- **`np.clip(arr, min_val, max_val)`:** Capping extreme outliers during EDA (e.g., ensuring satisfaction ratings stay within `[1, 5]`).

````python
exam_scores = np.array([45, 72, 88, 59, 95])
status = np.where(exam_scores >= 60, "Pass", "Fail")
print("Pass/Fail status:", status)

# Capping noisy survey ratings that accidentally exceeded [1, 5]
raw_ratings = np.array([0, 3, 5, 8, 2, -1, 4])
clean_ratings = np.clip(raw_ratings, 1, 5)
print("Clipped ratings [1, 5]:", clean_ratings)
````

### 8.2 Dot Product & Matrix Multiplication (`@` / `np.dot`)

#### 🌍 Real-World Use Case
**Deep learning and linear regression are all about matrix multiplication** — that is why knowing how `@` works is essential for ML!

Suppose you want to predict house prices using 2 features: **`[Size (sq ft), Bedrooms]`**.  
A simple linear model learns a weight for each feature (say `$200` per sq ft and `$15,000` per bedroom). Multiplying the **feature matrix `X`** `(3 houses × 2 features)` by the **weights vector `w`** `(2 features,)` computes all 3 predictions in a single step:

$$\text{Prediction} = (\text{Size} \times 200) + (\text{Bedrooms} \times 15000)$$

````python
# 3 houses, 2 features: [Size_sqft, Bedrooms]
X = np.array([
    [1000, 2],
    [1500, 3],
    [2200, 4]
])

# Learned weights: $200 per sq ft, $15,000 per bedroom
weights = np.array([200, 15_000])

# Matrix multiplication with @ (or np.dot(X, weights))
predicted_prices = X @ weights
print("Predicted house prices ($):", predicted_prices)
````

## 9. Capstone Mini-EDA Lab & Summary Cheat Sheet

### 🌍 Scenario: Cleaning & Preparing Store Branch Data
You receive a small raw dataset of **5 store branches (rows)** with **3 daily metrics (columns)**:  
`[Customers, Revenue ($), Rating (1-5)]`

Notice two real-world data issues inside `raw_data`:
1. Branch 2 is missing its `Revenue` (`np.nan`).
2. Branch 4 has an invalid `Rating` of `9.0` (ratings must be between `1` and `5`).

**Your Tasks:**
1. Inspect the `.shape` and count total `NaN` values.
2. Make a `.copy()` called `clean_data`, and replace the missing `Revenue` in column `1` with the mean revenue (`np.nanmean`) of the other branches.
3. Clip the `Rating` column (column `2`) so all ratings lie in `[1, 5]`.
4. Compute the **mean** and **std** of each metric (`axis=0`).
5. **Standardize (Z-score)** `clean_data` using broadcasting: `(clean_data - mean) / std`. *(Standardization is a classic ML preprocessing step so large numbers like Revenue don't dominate small numbers like Rating!)*

````python
raw_data = np.array([
    [120, 2400.0, 4.2],
    [ 85, 1750.0, 3.8],
    [150, np.nan, 4.7],
    [200, 4100.0, 4.5],
    [ 95, 1900.0, 9.0]  # Rating 9.0 is an invalid outlier!
])

# Inspect the .shape and count total NaN values.

print(f" 1. Shape is = {np.shape(raw_data)} , Nan count is = {np.isnan(raw_data).sum()}.")

# Make a .copy() called clean_data, and replace the missing Revenue in column 1 with the mean revenue (np.nanmean) of the other branches.

clean_data = raw_data.copy()
rev_mean = np.nanmean(clean_data[:, 1])
clean_data[:, 1] = np.where(np.isnan(clean_data[:, 1]), rev_mean, clean_data[:, 1])
print(f"2. cleaned data : {clean_data}")

# Clip the Rating column (column 2) so all ratings lie in [1, 5].

clean_data[:, 2] = np.clip(clean_data[:, 2], 1, 5)
print(f"3. cleaned data : {clean_data}")

# Compute the mean and std of each metric (axis=0).

col_mean = np.mean(clean_data, axis=0)
col_std = np.std(clean_data, axis=0)
print(f"4. Feature means: {np.round(col_mean, 2)}.")

# Standardize (Z-score) clean_data using broadcasting: (clean_data - mean) / std. (Standardization is a classic ML preprocessing step so large numbers like Revenue don't dominate small numbers like Rating!)

standardized = (clean_data - col_mean) / col_std
print(f"5. Standardized data (mean ≈ 0, std ≈ 1):  {np.round(standardized, 2)}.")
````
